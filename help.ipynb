```python
import pandas as pd
import numpy as np
```

```python
full_df_instances = pd.read_csv('instance_data_KS_25items_2026-05-08.csv')
```

```python
full_df_instances
```

```python
for i in range(4):
    for j in range(10):
        df_loc = full_df_instances[(full_df_instances['Type'] == i) & (full_df_instances['Instance'] == j)]
        print('Instance: ', i, 'Type: ', j, 'Count: ', len(df_loc))
        export_path = 'instance_data_KS_25items_Type_'+str(i)+'_Instance_'+str(j)+'.csv'
        df_loc.to_csv(export_path, index=False)
```
